**WAP to count the  number of unique vowels in a given string 
string= 'hai how are doing, what are you upto?**

```python
string9 = 'hai how are doing, what are you upto?'
count = 0
new = ''

for i in string9:
    if i in 'aeiou':
        if i not in new:   # check membership, not equality
            new = new + i
            count = count + 1

print("The number of unique vowels are:", count)
print("Unique vowels are:", new)
  
```

```python
string9 = 'hai how are doing, what are you upto?'
count = 0
new = ''

for i in string9:
    if i in 'aeiou':
        unique = True
        # compare i with each character in new
        for j in new:
            if i == j:      # if already present
                unique = False
                break
        if unique:          # only add if not found
            new = new + i
            count = count + 1

print("The number of unique vowels are:", count)
print("Unique vowels are:", new)

             
```

```python
string9 = 'hai how are doing, what are you upto?'
count = 0
new = ''

for i in string9:
    if i in 'aeiou':
        # assume it's unique until proven otherwise
        for j in new:
            if i == j:   # found duplicate
                break
        else:            # this else belongs to the for-loop, runs only if no break
            new = new + i
            count = count + 1

print("The number of unique vowels are:", count)
print("Unique vowels are:", new)
```
